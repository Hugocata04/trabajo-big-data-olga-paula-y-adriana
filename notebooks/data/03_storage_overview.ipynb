# Capa de Almacenamiento – Modelo Data Lake

## RAW
- CSV originales sin modificar.

## PROCESSED
- Datos limpios en formato **Parquet** (optimizado para Spark).

## ANALYTICS
- Tablas agregadas para dashboards.

Este proyecto sigue un modelo tipo **Data Lake por zonas**:
- Raw → Ingreso
- Processed → Limpieza + estandarización (Parquet)
- Analytics → Datos enriquecidos/listos para consumo
