## 1. Análisis de la actividad del cliente


Un **KPI de actividad** es un indicador que mide qué tanto y con qué frecuencia un cliente utiliza un producto, servicio o plataforma. Estos KPIs permiten evaluar el nivel de uso real, detectar clientes inactivos y entender los patrones de interacción a lo largo del tiempo. Son fundamentales para conocer el engagement operativo del cliente y anticipar riesgos de abandono. 
Se generan variables como:
- **Frecuencia de actividades por cliente**: Mide el número total de acciones realizadas por cada cliente; sirve para identificar usuarios muy activos frente a usuarios con poco uso.
- **Días desde la última actividad (Recencia)**: Indica cuánto tiempo ha pasado desde que el cliente usó el servicio por última vez; cuanto mayor es este valor, mayor riesgo de abandono.
- **Nivel de actividad (Baja / Media / Alta)**: Clasifica al cliente según su uso reciente y acumulado del sistema, ofreciendo una visión rápida de su nivel de engagement.


```python
# 1. Estudiar la actividad y la frecuencia de los clientes

# Crear columnas de tiempo
df_behaviour = df_behaviour.withColumn("YEAR", year(col("DATE"))) \
                           .withColumn("MONTH", month(col("DATE"))) \
                           .withColumn("YEAR_MONTH", date_format(col("DATE"), "yyyy-MM"))

# Calcular recencia (días desde última actividad) para medir la inactividad:

# Clientes con recencia baja = más propensos a comprar de nuevo
# Cliente con alta recencia = riesgo de abandono 

ultima_fecha = df_behaviour.agg(F.max("DATE")).collect()[0][0]
df_behaviour = df_behaviour.withColumn("DAYS_SINCE_ACTIVITY", 
                                       datediff(lit(ultima_fecha), col("DATE")))


# Frecuencia de transacciones por cliente
kpi_frecuencia = df_behaviour.groupBy("CLIENT_ID").agg(
    count("*").alias("TOTAL_TRANSACCIONES"),
    countDistinct("DATE").alias("DIAS_CON_ACTIVIDAD"),
    mean("CREDIT_CARD_DRAWINGS").alias("GASTO_PROMEDIO_DIARIO"),
    F.stddev("CREDIT_CARD_DRAWINGS").alias("GASTO_VARIABILIDAD")
)

# Recencia (RFM - Recency)
kpi_recencia = df_behaviour.groupBy("CLIENT_ID").agg(
    F.min("DAYS_SINCE_ACTIVITY").alias("DIAS_DESDE_ULTIMA_ACTIVIDAD")
)

# Para ver si un cliente está activo, moderado, poco activo o inactivo:

# Unir KPIs de actividad
kpi_actividad = kpi_frecuencia.join(kpi_recencia, on="CLIENT_ID", how="inner")

# Segmentación por actividad
kpi_actividad = kpi_actividad.withColumn("SEGMENTO_ACTIVIDAD",
    when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") <= 30, "ACTIVO")
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") <= 90, "MODERADO")
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") <= 180, "POCO_ACTIVO")
    .otherwise("INACTIVO")
)

print(f"KPIs de actividad calculados: {kpi_actividad.count():,} clientes")
```

## 2. Análisis del valor económico

Un **KPI de valor económico** mide la contribución financiera que un cliente aporta al negocio, ya sea mediante ingresos, transacciones o rentabilidad. Estos indicadores permiten identificar a los clientes más importantes en términos económicos y priorizar acciones comerciales según su impacto en el negocio.
Este kpi incluye:
- **Valor total generado por cliente**: Suma de todos los montos o ingresos asociados al cliente; permite identificar a los clientes más valiosos.
- **Rentabilidad estimada**: Mide el beneficio neto que aporta el cliente considerando ingresos y costos asociados; ayuda a detectar clientes realmente rentables.
- **Segmentación económica**: Agrupa a los clientes en categorías como Alto, Medio o Bajo valor para facilitar el análisis y la toma de decisiones comerciales.

```python
# 2. KPI de valor económico y rentabilidad


# Verificar columnas disponibles en df_behaviour
print("Columnas en df_behaviour:", df_behaviour.columns)

# Identificar la columna correcta para el balance
balance_col = None
for col_name in df_behaviour.columns:
    if 'balance' in col_name.lower() or 'balan' in col_name.lower():
        balance_col = col_name
        print(f" Columna de balance encontrada: {balance_col}")
        break


# Comportamiento financiero
kpi_financiero = df_behaviour.groupBy("CLIENT_ID").agg(
    F.sum("CREDIT_CARD_DRAWINGS").alias("GASTO_TOTAL"),
    F.avg(balance_col).alias("SALDO_PROMEDIO"),
    F.max("CREDIT_CARD_LIMIT").alias("LIMITE_MAXIMO"),
    (F.sum("CREDIT_CARD_DRAWINGS") / F.count("*")).alias("GASTO_POR_TRANSACCION")
)

# Calcular uso del límite (%)
kpi_financiero = kpi_financiero.withColumn(
    "PORCENTAJE_USO_LIMITE",
    F.when(F.col("LIMITE_MAXIMO") > 0,
           (F.col("SALDO_PROMEDIO") / F.col("LIMITE_MAXIMO")) * 100)
    .otherwise(0)
)

# Información de clientes
kpi_clientes = df_clients.select(
    "CLIENT_ID",
    "TOTAL_INCOME",
    "AMOUNT_PRODUCT",
    "NUMBER_OF_PRODUCTS",
    "NAME_PRODUCT_TYPE"
)

# Calcular rentabilidad estimada
kpi_valor = kpi_financiero.join(kpi_clientes, on="CLIENT_ID", how="inner")
kpi_valor = kpi_valor.withColumn(
    "RENTABILIDAD_ESTIMADA",
    F.col("TOTAL_INCOME") - F.col("GASTO_TOTAL")
)

# Segmentación por valor 
window_spec = Window.orderBy(F.col("GASTO_TOTAL").desc())
kpi_valor = kpi_valor.withColumn("RANK_VALOR", F.percent_rank().over(window_spec))

kpi_valor = kpi_valor.withColumn("SEGMENTO_VALOR",
    F.when(F.col("RANK_VALOR") <= 0.2, "A - ALTO VALOR")
    .when(F.col("RANK_VALOR") <= 0.5, "B - VALOR MEDIO")
    .otherwise("C - BAJO VALOR")
)

print(f" KPIs de valor creados: {kpi_valor.count():,} clientes")
print(" Columnas creadas:", kpi_valor.columns)
```

Con tal de facilitar la comprensión de lo que se acaba de calcular, se realizan una serie de gráficas que permiten visualizar los resultados.

```python
# Visualizaciones de KPIs de valor económico

# Convertir a Pandas para visualización

kpi_valor_pd = kpi_valor.toPandas()

fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=('Segmentos', 'Gasto', 'Rentabilidad', 'Uso Límite', 'Productos', 'Top 10'),
    specs=[
        [{'type': 'domain'}, {'type': 'xy'}, {'type': 'xy'}],
        [{'type': 'xy'}, {'type': 'xy'}, {'type': 'xy'}]
    ]
)

# 1. Tarta
valor_counts = kpi_valor_pd['SEGMENTO_VALOR'].value_counts()
fig.add_trace(go.Pie(labels=valor_counts.index, values=valor_counts.values), row=1, col=1)

# 2. Barras
gasto_segmento = kpi_valor_pd.groupby('SEGMENTO_VALOR')['GASTO_TOTAL'].mean()
fig.add_trace(go.Bar(x=gasto_segmento.index, y=gasto_segmento.values), row=1, col=2)

# 3. Dispersión 
n_points = 500 if len(kpi_valor_pd) > 500 else len(kpi_valor_pd)
sample = kpi_valor_pd.sample(n=n_points)
fig.add_trace(go.Scatter(x=sample['TOTAL_INCOME'], y=sample['RENTABILIDAD_ESTIMADA'], mode='markers'), row=1, col=3)

# 4. Histograma
fig.add_trace(go.Histogram(x=kpi_valor_pd['PORCENTAJE_USO_LIMITE']), row=2, col=1)

# 5. Boxplot
fig.add_trace(go.Box(y=kpi_valor_pd['NUMBER_OF_PRODUCTS']), row=2, col=2)

# 6. Top 10
top_10 = kpi_valor_pd.nlargest(10, 'RENTABILIDAD_ESTIMADA')
fig.add_trace(go.Bar(x=top_10['CLIENT_ID'].astype(str), y=top_10['RENTABILIDAD_ESTIMADA']), row=2, col=3)

fig.update_layout(height=800, title_text="Valor Económico")
fig.show()
```

## 3. Análisis de interacción y fidelidad

Un **KPI de fidelidad e interacción** evalúa el nivel de compromiso, conexión y continuidad del cliente con la empresa, basado en la recurrencia, consistencia y calidad de sus interacciones. Permite medir qué tan vinculado está el cliente y anticipar comportamientos futuros relacionados con retención o abandono.
Se construyen métricas como:
- **Número de interacciones**: Cuenta todas las veces que el cliente se relaciona con la empresa; mayor número implica mayor compromiso.
- **Consistencia del uso**: Evalúa si las interacciones se mantienen en el tiempo o son esporádicas; un uso constante refleja mayor fidelidad.
- **Engagement**: Indicador global que combina frecuencia, recencia y consistencia para medir el compromiso del cliente.
- **Segmento de fidelidad**: Clasificación final del cliente (Alta, Media, Baja fidelidad) según sus patrones de interacción y compromiso.


```python
#  3. KPIs de interacción y fidelidad

# Interacción digital
kpi_digital = df_clients.select(
    "CLIENT_ID",
    "DIGITAL_CLIENT",
    "HOME_OWNER",
    "OWN_INSURANCE_CAR"
).withColumn(
    "CLIENTE_DIGITAL",
    when(col("DIGITAL_CLIENT") == 1, "SI").otherwise("NO")
)

# Comportamiento de fidelidad (retención)
# Calcular meses consecutivos con actividad
window_cliente = Window.partitionBy("CLIENT_ID").orderBy("YEAR", "MONTH")

# Primero agrupar por cliente y mes
df_mensual = df_behaviour.groupBy("CLIENT_ID", "YEAR", "MONTH", "YEAR_MONTH").agg(
    count("*").alias("TRANSACCIONES_MES")
).withColumn(
    "MES_ANTERIOR",
    lag("YEAR_MONTH", 1).over(window_cliente)
).withColumn(
    "ES_CONSECUTIVO",
    when(
        (col("YEAR") == lag("YEAR", 1).over(window_cliente)) & 
        (col("MONTH") - lag("MONTH", 1).over(window_cliente) == 1), 
        1
    ).otherwise(0)
)

# Calcular racha máxima de meses consecutivos
kpi_racha = df_mensual.groupBy("CLIENT_ID").agg(
    F.sum("ES_CONSECUTIVO").alias("MESES_CONSECUTIVOS"),
    count("*").alias("MESES_TOTALES")
).withColumn(
    "TASA_RETENCION",
    when(col("MESES_TOTALES") > 0,
         (col("MESES_CONSECUTIVOS") / col("MESES_TOTALES")) * 100)
    .otherwise(0)
)

# Unir todos los KPIs de fidelidad
kpi_fidelidad_final = kpi_digital.join(kpi_racha, on="CLIENT_ID", how="left")

# Segmentación por fidelidad
kpi_fidelidad_final = kpi_fidelidad_final.withColumn("SEGMENTO_FIDELIDAD",
    when(col("TASA_RETENCION") >= 80, "ALTA_FIDELIDAD")
    .when(col("TASA_RETENCION") >= 50, "MEDIA_FIDELIDAD")
    .otherwise("BAJA_FIDELIDAD")
).fillna({"TASA_RETENCION": 0, "SEGMENTO_FIDELIDAD": "SIN_HISTORIAL"})

print(f" KPIs de fidelidad calculados: {kpi_fidelidad_final.count():,} clientes")
```

## 4. Análisis de riesgo potencial

Un **KPI de riesgo** evalúa la probabilidad de incumplimiento financiero y de abandono (churn) del cliente, basado en su comportamiento crediticio, situación económica y patrones de actividad reciente. Permite medir el nivel de riesgo asociado a cada cliente, clasificarlo según su perfil de riesgo y anticipar posibles pérdidas o desvinculaciones, facilitando la toma de decisiones preventivas y la asignación de recursos.

Se construyen métricas como:

- **Score promedio:** Promedio de puntuaciones de riesgo (reactiva, proactiva y conductual) que refleja la solvencia general del cliente.
- **Ratio deuda/ingreso:** Porcentaje que representa la carga de deuda en relación con los ingresos totales; un valor elevado indica mayor vulnerabilidad financiera.
- **Probabilidad de abandono:** Estimación de la tendencia al churn, calculada a partir de la inactividad reciente y la tasa de retención histórica.
- **Nivel de riesgo:** Clasificación final del cliente (Alto, Medio, Bajo riesgo, Riesgo financiero, Riesgo contrato) según su probabilidad de abandono, incumplimiento y sobreendeudamiento.

```python
# 4.  KPIs de riesgo

# Riesgo financiero
kpi_riesgo = df_clients.select(
    "CLIENT_ID",
    "REACTIVE_SCORING",
    "PROACTIVE_SCORING", 
    "BEHAVIORAL_SCORING",
    "NON_COMPLIANT_CONTRACT",
    "INSTALLMENT",
    "TOTAL_INCOME"
)

# Calcular score promedio y ratios de riesgo
kpi_riesgo = kpi_riesgo.withColumn(
    "SCORE_PROMEDIO",
    when(
        (col("REACTIVE_SCORING").isNotNull()) &
        (col("PROACTIVE_SCORING").isNotNull()) &
        (col("BEHAVIORAL_SCORING").isNotNull()),
        (col("REACTIVE_SCORING") + col("PROACTIVE_SCORING") + col("BEHAVIORAL_SCORING")) / 3
    ).otherwise(0)
).withColumn(
    "RATIO_DEUDA_INGRESO",
    when(col("TOTAL_INCOME") > 0, 
         (col("INSTALLMENT") / col("TOTAL_INCOME")) * 100)
    .otherwise(100)
).fillna({"SCORE_PROMEDIO": 0, "RATIO_DEUDA_INGRESO": 100})

# Riesgo de abandono  - unir con actividad y fidelidad
kpi_churn = kpi_actividad.select(
    "CLIENT_ID",
    "DIAS_DESDE_ULTIMA_ACTIVIDAD",
    "SEGMENTO_ACTIVIDAD"
).join(
    kpi_fidelidad_final.select("CLIENT_ID", "TASA_RETENCION"),
    on="CLIENT_ID",
    how="inner"
)

# Calcular probabilidad de abandono
kpi_churn = kpi_churn.withColumn("PROBABILIDAD_ABANDONO",
    when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") > 180, 0.9)
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") > 90, 0.7)
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") > 30, 0.4)
    .when(col("TASA_RETENCION") < 30, 0.6)
    .otherwise(0.2)
)

# Unir todos los KPIs de riesgo
kpi_riesgo_final = kpi_riesgo.join(kpi_churn, on="CLIENT_ID", how="inner")

# Segmentación por riesgo
kpi_riesgo_final = kpi_riesgo_final.withColumn("NIVEL_RIESGO",
    when(col("PROBABILIDAD_ABANDONO") >= 0.7, "ALTO_RIESGO")
    .when(col("PROBABILIDAD_ABANDONO") >= 0.4, "MEDIO_RIESGO")
    .when(col("NON_COMPLIANT_CONTRACT") == 1, "RIESGO_CONTRATO")
    .when(col("RATIO_DEUDA_INGRESO") > 40, "RIESGO_FINANCIERO")
    .otherwise("BAJO_RIESGO")
)

print(f" KPIs de riesgo calculados: {kpi_riesgo_final.count():,} clientes")
```

## 5. Análisis de oportunidades comerciales

Un **KPI de oportunidades comerciales** identifica clientes con alto potencial para acciones de venta adicional, reactivación o migración, basándose en su perfil de valor, actividad, fidelidad y riesgo. Permite priorizar iniciativas comerciales dirigidas a maximizar el valor del cliente, incrementar la penetración de productos y fortalecer la relación mediante ofertas personalizadas.

Se construyen métricas como:

- **Oportunidad de cross-selling:** Clasifica a clientes con pocos productos pero alto valor o ingresos, identificando potencial para ampliar la cartera de productos contratados.
- **Oportunidad de engagement (re-engagement):** Detecta clientes inactivos de alto valor, no digitalizados o sin seguros activos, señalando oportunidades para recuperación, migración digital o venta de seguros.
- **Oportunidad de up-selling:** Identifica clientes con alta utilización de límite crediticio y buen score, o con rentabilidad positiva y pocos productos, sugiriendo opciones para aumentar límites o ofrecer productos premium.

```python
# 5. KPI de oportunidades comerciales

# Unir todos los KPIs para análisis cruzado
kpi_completo = kpi_actividad \
    .join(kpi_valor, on="CLIENT_ID", how="inner") \
    .join(kpi_fidelidad_final.select("CLIENT_ID", "CLIENTE_DIGITAL", "TASA_RETENCION", "OWN_INSURANCE_CAR"), 
          on="CLIENT_ID", how="inner") \
    .join(kpi_riesgo_final.select("CLIENT_ID", "SCORE_PROMEDIO", "RATIO_DEUDA_INGRESO", "NON_COMPLIANT_CONTRACT"), 
          on="CLIENT_ID", how="inner")

# Añadir CAR_AGE desde df_clients si existe
if "CAR_AGE" in df_clients.columns:
    car_age_df = df_clients.select("CLIENT_ID", "CAR_AGE")
    kpi_completo = kpi_completo.join(car_age_df, on="CLIENT_ID", how="left")
else:
    kpi_completo = kpi_completo.withColumn("CAR_AGE", lit(None))

# KPI 5.1: Oportunidades de cross-selling
kpi_oportunidades = kpi_completo.withColumn("OPORTUNIDAD_CROSS_SELLING",
    when((col("NUMBER_OF_PRODUCTS") == 1) & 
         (col("SEGMENTO_VALOR").isin(["A - ALTO VALOR", "B - VALOR MEDIO"])), 
         "ALTA_OPORTUNIDAD")
    .when((col("NUMBER_OF_PRODUCTS") <= 2) & 
          (col("TOTAL_INCOME") > kpi_completo.agg(mean("TOTAL_INCOME")).collect()[0][0]), 
          "MEDIA_OPORTUNIDAD")
    .otherwise("BAJA_OPORTUNIDAD")
)

# KPI 5.2: Oportunidades de engagement (re-engagement)
kpi_oportunidades = kpi_oportunidades.withColumn("OPORTUNIDAD_ENGAGEMENT",
    when((col("SEGMENTO_ACTIVIDAD") == "INACTIVO") & 
         (col("SEGMENTO_VALOR") == "A - ALTO VALOR"), 
         "RECUPERACION_ALTO_VALOR")
    .when((col("SEGMENTO_ACTIVIDAD").isin(["POCO_ACTIVO", "INACTIVO"])) & 
          (col("CLIENTE_DIGITAL") == "NO"), 
          "MIGRACION_DIGITAL")
    .when((col("OWN_INSURANCE_CAR") == "N") & 
          (col("CAR_AGE").isNotNull()) & (col("CAR_AGE") > 0), 
          "SEGURO_AUTOMOVIL")
    .otherwise("MANTENIMIENTO")
)

# KPI 5.3: Clientes ideales para up-selling
kpi_oportunidades = kpi_oportunidades.withColumn("OPORTUNIDAD_UP_SELLING",
    when((col("PORCENTAJE_USO_LIMITE") > 70) & 
         (col("SCORE_PROMEDIO") > 70), 
         "AUMENTO_LIMITE_CREDITO")
    .when((col("RENTABILIDAD_ESTIMADA") > 0) & 
          (col("NUMBER_OF_PRODUCTS") < 3), 
          "PRODUCTOS_PREMIUM")
    .otherwise("NO_APLICA")
)

print(f"KPIs de oportunidades calculados: {kpi_oportunidades.count():,} clientes")
```

```python
# Visualización de oportunidades comerciales
kpi_oportunidades_pd = kpi_oportunidades.toPandas()

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Oportunidades de Cross-Selling',
                    'Oportunidades de Engagement',
                    'Segmentación por Valor y Actividad',
                    'Clientes Ideales para Up-Selling'),
    specs=[[{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'scatter'}, {'type': 'bar'}]]
)

# Oportunidades cross-selling
cross_selling_counts = kpi_oportunidades_pd['OPORTUNIDAD_CROSS_SELLING'].value_counts()
fig.add_trace(
    go.Bar(x=cross_selling_counts.index, 
           y=cross_selling_counts.values,
           name='Cross-Selling',
           marker_color=['green', 'orange', 'lightblue']),
    row=1, col=1
)

# Oportunidades engagement
engagement_counts = kpi_oportunidades_pd['OPORTUNIDAD_ENGAGEMENT'].value_counts()
fig.add_trace(
    go.Bar(x=engagement_counts.index, 
           y=engagement_counts.values,
           name='Engagement',
           marker_color='lightcoral'),
    row=1, col=2
)

# Dispersión valor vs actividad (para identificar oportunidades)
fig.add_trace(
    go.Scatter(x=kpi_oportunidades_pd['TOTAL_TRANSACCIONES'],
               y=kpi_oportunidades_pd['GASTO_TOTAL'],
               mode='markers',
               marker=dict(
                   size=10,
                   color=kpi_oportunidades_pd['SEGMENTO_ACTIVIDAD'].map({
                       'ACTIVO': 'green',
                       'MODERADO': 'yellow',
                       'POCO_ACTIVO': 'orange',
                       'INACTIVO': 'red'
                   }).fillna('gray'),
                   opacity=0.7
               ),
               text=kpi_oportunidades_pd['OPORTUNIDAD_CROSS_SELLING'],
               hoverinfo='text+x+y',
               name='Clientes'),
    row=2, col=1
)

# Oportunidades up-selling
up_selling_counts = kpi_oportunidades_pd['OPORTUNIDAD_UP_SELLING'].value_counts()
fig.add_trace(
    go.Bar(x=up_selling_counts.index, 
           y=up_selling_counts.values,
           name='Up-Selling',
           marker_color=['gold', 'silver', 'lightgray']),
    row=2, col=2
)

fig.update_layout(
    height=800, 
    title_text=" KPIs de Oportunidades Comerciales - FinPlus",
    template='plotly_white'
)

# Mejorar ejes
fig.update_xaxes(title_text="Tipo Oportunidad", row=1, col=1)
fig.update_xaxes(title_text="Tipo Oportunidad", row=1, col=2)
fig.update_xaxes(title_text="Total Transacciones", row=2, col=1)
fig.update_xaxes(title_text="Tipo Oportunidad", row=2, col=2)

fig.update_yaxes(title_text="Número de Clientes", row=1, col=1)
fig.update_yaxes(title_text="Número de Clientes", row=1, col=2)
fig.update_yaxes(title_text="Gasto Total (€)", row=2, col=1)
fig.update_yaxes(title_text="Número de Clientes", row=2, col=2)

fig.show()
```

```python
# Exportar KPIs

import os
import pandas as pd
import json
from pyspark.sql import Row

# Crear directorio de salida si no existe
output_dir = "/home/jovyan/work/finplus_data/kpi_results/"
os.makedirs(output_dir, exist_ok=True)

# Verificar qué variables tenemos disponibles
print("Variables disponibles:")
variables_disponibles = []
for var_name in dir():
    if not var_name.startswith('_'):
        variables_disponibles.append(var_name)
print(", ".join(variables_disponibles[:20]), "...")

# 1. Exportar KPIs completos 
if 'kpi_oportunidades' in globals():
    try:
        kpi_oportunidades.write \
            .mode("overwrite") \
            .option("header", "true") \
            .parquet(output_dir + "kpis_completos.parquet")
        print(f"✓ KPIs completos exportados: {output_dir}kpis_completos.parquet")
    except Exception as e:
        print(f"✗ Error exportando kpis completos: {e}")
elif 'df_base' in globals():
    try:
        df_base.write \
            .mode("overwrite") \
            .option("header", "true") \
            .parquet(output_dir + "kpis_completos.parquet")
        print(f"✓ DataFrame base exportado: {output_dir}kpis_completos.parquet")
    except Exception as e:
        print(f"✗ Error exportando DataFrame base: {e}")
else:
    print("✗ No se encontró ningún DataFrame para exportar como kpis completos")

# 2. Exportar resumen para Tableau (CSV) 
if 'resumen_kpis' in globals() and isinstance(resumen_kpis, dict):
    try:
        # Convertir el diccionario a DataFrame de Spark
        resumen_df = spark.createDataFrame([Row(**resumen_kpis)])
        resumen_df.write \
            .mode("overwrite") \
            .option("header", "true") \
            .csv(output_dir + "resumen_kpis.csv")
        print(f"✓ Resumen de KPIs exportado: {output_dir}resumen_kpis.csv")
    except Exception as e:
        print(f"✗ Error exportando resumen CSV: {e}")
        
        # Intentar guardar como pandas DataFrame
        try:
            resumen_pd = pd.DataFrame([resumen_kpis])
            resumen_pd.to_csv(output_dir + "resumen_kpis.csv", index=False)
            print(f"✓ Resumen exportado usando pandas: {output_dir}resumen_kpis.csv")
        except Exception as e2:
            print(f"✗ Error también con pandas: {e2}")
else:
    print("✗ No se encontró diccionario resumen_kpis")

# 3. Exportar datos para visualización 
dataframe_para_visualizacion = None
if 'kpi_oportunidades' in globals():
    dataframe_para_visualizacion = kpi_oportunidades
elif 'df_base' in globals():
    dataframe_para_visualizacion = df_base
elif 'kpi_riesgo_final' in globals():
    dataframe_para_visualizacion = kpi_riesgo_final

if dataframe_para_visualizacion is not None:
    try:
        # Verificar qué columnas están disponibles
        columnas_disponibles = dataframe_para_visualizacion.columns
        print(f"Columnas disponibles: {', '.join(columnas_disponibles[:10])}...")
        
        # Seleccionar columnas comunes para visualización
        columnas_seleccionadas = []
        posibles_columnas = [
            "CLIENT_ID", "CLIENT_ID", "ID", "id", "customer_id",
            "SEGMENTO_ACTIVIDAD", "SEGMENTO_VALOR", "SEGMENTO_FIDELIDAD",
            "NIVEL_RIESGO", "TOTAL_TRANSACCIONES", "GASTO_TOTAL",
            "RENTABILIDAD_ESTIMADA", "TASA_RETENCION", "PORCENTAJE_USO_LIMITE",
            "OPORTUNIDAD_CROSS_SELLING", "OPORTUNIDAD_ENGAGEMENT",
            "PROBABILIDAD_ABANDONO", "SCORE_PROMEDIO", "RATIO_DEUDA_INGRESO"
        ]
        
        for col in posibles_columnas:
            if col in columnas_disponibles:
                columnas_seleccionadas.append(col)
        
        # Limitar el número de registros para visualización
        kpi_visualizacion = dataframe_para_visualizacion.select(columnas_seleccionadas).limit(100000)
        
        # Escribir a CSV
        kpi_visualizacion.write \
            .mode("overwrite") \
            .option("header", "true") \
            .csv(output_dir + "kpis_visualizacion.csv")
        
        print(f"✓ KPIs para visualización exportados: {output_dir}kpis_visualizacion.csv")
        print(f"  Columnas incluidas: {', '.join(columnas_seleccionadas[:10])}...")
        
    except Exception as e:
        print(f"✗ Error exportando datos para visualización: {e}")
else:
    print("✗ No se encontró DataFrame para exportar datos de visualización")

# 4. Guardar también en formato JSON para fácil acceso
if 'resumen_kpis' in globals() and isinstance(resumen_kpis, dict):
    try:
        with open(output_dir + "resumen_kpis.json", "w") as f:
            json.dump(resumen_kpis, f, indent=2, default=str)
        print(f"✓ Resumen en JSON exportado: {output_dir}resumen_kpis.json")
    except Exception as e:
        print(f"✗ Error exportando JSON: {e}")
        
        # Intentar con serialización manual
        try:
            # Convertir valores no serializables a string
            resumen_serializable = {}
            for k, v in resumen_kpis.items():
                try:
                    json.dumps({k: v})
                    resumen_serializable[k] = v
                except:
                    resumen_serializable[k] = str(v)
            
            with open(output_dir + "resumen_kpis.json", "w") as f:
                json.dump(resumen_serializable, f, indent=2)
            print(f"✓ Resumen en JSON exportado (conversión manual): {output_dir}resumen_kpis.json")
        except Exception as e2:
            print(f"✗ Error incluso con conversión manual: {e2}")

# 5. Crear archivo README con descripción
try:
    # Obtener total de clientes si existe
    total_clientes = "N/A"
    if 'resumen_kpis' in globals() and 'TOTAL_CLIENTES' in resumen_kpis:
        total_clientes = f"{resumen_kpis['TOTAL_CLIENTES']:,}"
    elif 'resumen_kpis' in globals() and 'CLIENTES_TOTALES' in resumen_kpis:
        total_clientes = f"{resumen_kpis['CLIENTES_TOTALES']:,}"
    
    readme_content = f"""
# KPIs FINPLUS ANALYTICS CHALLENGE

## Archivos generados:
1. **kpis_completos.parquet** - Todos los KPIs calculados para cada cliente
2. **resumen_kpis.csv** - Resumen ejecutivo de métricas agregadas
3. **kpis_visualizacion.csv** - Subconjunto para dashboards (hasta 100k registros)
4. **resumen_kpis.json** - Resumen en formato JSON

## KPIs calculados:
- Actividad: Frecuencia, recencia, segmentación
- Valor económico: Gasto, rentabilidad, segmentación ABC
- Fidelidad: Retención, digitalización
- Riesgo: Scoring, probabilidad de churn
- Oportunidades: Cross-selling, engagement, up-selling

## Total clientes analizados: {total_clientes}

Generado el: {pd.Timestamp.now().strftime('%Y-%m-%d %H:%M:%S')}
"""

    with open(output_dir + "README.md", "w") as f:
        f.write(readme_content)
    
    print(f"✓ README.md creado: {output_dir}README.md")
    
except Exception as e:
    print(f"✗ Error creando README: {e}")

# 6. Mostrar métricas finales si están disponibles
print("MÉTRICAS FINALES DISPONIBLES PARA EL DASHBOARD:")

if 'resumen_kpis' in globals() and isinstance(resumen_kpis, dict):
    # Definir métricas clave con valores por defecto
    metricas_clave = {
        'Clientes Totales': ['TOTAL_CLIENTES', 'CLIENTES_TOTALES'],
        'Clientes Activos': ['CLIENTES_ACTIVOS'],
        'Clientes Digitales': ['CLIENTES_DIGITALES'],
        'Clientes Alto Valor': ['CLIENTES_ALTO_VALOR'],
        'Clientes Alto Riesgo': ['CLIENTES_ALTO_RIESGO'],
        'Gasto Promedio': ['GASTO_TOTAL_PROMEDIO'],
        'Rentabilidad Promedio': ['RENTABILIDAD_PROMEDIO'],
        'Tasa Retención': ['TASA_RETENCION_PROMEDIO', 'TASA_RETENCION'],
        'Oportunidades Cross-Selling': ['OPORTUNIDADES_CROSS_SELLING'],
        'Oportunidades Re-engagement': ['OPORTUNIDADES_ENGAGEMENT']
    }
    
    for nombre_metric, posibles_keys in metricas_clave.items():
        valor = "N/A"
        for key in posibles_keys:
            if key in resumen_kpis and resumen_kpis[key] is not None:
                if isinstance(resumen_kpis[key], (int, float)):
                    if nombre_metric in ['Gasto Promedio', 'Rentabilidad Promedio']:
                        valor = f"€{resumen_kpis[key]:,.2f}"
                    elif nombre_metric == 'Tasa Retención':
                        valor = f"{resumen_kpis[key]:.1f}%"
                    else:
                        valor = f"{resumen_kpis[key]:,}"
                else:
                    valor = str(resumen_kpis[key])
                break
        
        print(f"{nombre_metric:30} {valor:>20}")
else:
    print("No se encontraron métricas para mostrar")


# 7. Verificar archivos creados
print("\n✓ Verificación de archivos creados:")
archivos_creados = []
for archivo in os.listdir(output_dir):
    if archivo.endswith(('.parquet', '.csv', '.json', '.md')):
        archivos_creados.append(archivo)

if archivos_creados:
    for archivo in archivos_creados:
        ruta_completa = os.path.join(output_dir, archivo)
        tamaño = os.path.getsize(ruta_completa) if os.path.exists(ruta_completa) else 0
        print(f"  - {archivo} ({tamaño:,} bytes)")
else:
    print("  No se crearon archivos en el directorio de salida")

print(f"\n✓ Proceso de exportación completado en: {output_dir}")
```
