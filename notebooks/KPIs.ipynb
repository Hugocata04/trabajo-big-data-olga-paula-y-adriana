```python
# 1. Estudiar la actividad y la frecuencia de los clientes

# Crear columnas de tiempo
df_behaviour = df_behaviour.withColumn("YEAR", year(col("DATE"))) \
                           .withColumn("MONTH", month(col("DATE"))) \
                           .withColumn("YEAR_MONTH", date_format(col("DATE"), "yyyy-MM"))

# Calcular recencia (días desde última actividad) para medir la inactividad:

# Clientes con recencia baja = más propensos a comprar de nuevo
# Cliente con alta recencia = riesgo de abandono 

ultima_fecha = df_behaviour.agg(F.max("DATE")).collect()[0][0]
df_behaviour = df_behaviour.withColumn("DAYS_SINCE_ACTIVITY", 
                                       datediff(lit(ultima_fecha), col("DATE")))


# 1. Frecuencia de transacciones por cliente
kpi_frecuencia = df_behaviour.groupBy("CLIENT_ID").agg(
    count("*").alias("TOTAL_TRANSACCIONES"),
    countDistinct("DATE").alias("DIAS_CON_ACTIVIDAD"),
    mean("CREDIT_CARD_DRAWINGS").alias("GASTO_PROMEDIO_DIARIO"),
    F.stddev("CREDIT_CARD_DRAWINGS").alias("GASTO_VARIABILIDAD")
)

# 2. Recencia (RFM - Recency)
kpi_recencia = df_behaviour.groupBy("CLIENT_ID").agg(
    F.min("DAYS_SINCE_ACTIVITY").alias("DIAS_DESDE_ULTIMA_ACTIVIDAD")
)

# Para ver si un cliente está activo, moderado, poco activo o inactivo:

# Unir KPIs de actividad
kpi_actividad = kpi_frecuencia.join(kpi_recencia, on="CLIENT_ID", how="inner")

# Segmentación por actividad
kpi_actividad = kpi_actividad.withColumn("SEGMENTO_ACTIVIDAD",
    when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") <= 30, "ACTIVO")
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") <= 90, "MODERADO")
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") <= 180, "POCO_ACTIVO")
    .otherwise("INACTIVO")
)

print(f"KPIs de actividad calculados: {kpi_actividad.count():,} clientes")
```

```python
# 2. KPI de valor económico y rentabilidad


# Verificar columnas disponibles en df_behaviour
print("Columnas en df_behaviour:", df_behaviour.columns)

# Identificar la columna correcta para el balance
balance_col = None
for col_name in df_behaviour.columns:
    if 'balance' in col_name.lower() or 'balan' in col_name.lower():
        balance_col = col_name
        print(f" Columna de balance encontrada: {balance_col}")
        break


# Comportamiento financiero
kpi_financiero = df_behaviour.groupBy("CLIENT_ID").agg(
    F.sum("CREDIT_CARD_DRAWINGS").alias("GASTO_TOTAL"),
    F.avg(balance_col).alias("SALDO_PROMEDIO"),
    F.max("CREDIT_CARD_LIMIT").alias("LIMITE_MAXIMO"),
    (F.sum("CREDIT_CARD_DRAWINGS") / F.count("*")).alias("GASTO_POR_TRANSACCION")
)

# Calcular uso del límite (%)
kpi_financiero = kpi_financiero.withColumn(
    "PORCENTAJE_USO_LIMITE",
    F.when(F.col("LIMITE_MAXIMO") > 0,
           (F.col("SALDO_PROMEDIO") / F.col("LIMITE_MAXIMO")) * 100)
    .otherwise(0)
)

# Información de clientes
kpi_clientes = df_clients.select(
    "CLIENT_ID",
    "TOTAL_INCOME",
    "AMOUNT_PRODUCT",
    "NUMBER_OF_PRODUCTS",
    "NAME_PRODUCT_TYPE"
)

# Calcular rentabilidad estimada
kpi_valor = kpi_financiero.join(kpi_clientes, on="CLIENT_ID", how="inner")
kpi_valor = kpi_valor.withColumn(
    "RENTABILIDAD_ESTIMADA",
    F.col("TOTAL_INCOME") - F.col("GASTO_TOTAL")
)

# 4. Segmentación por valor 
window_spec = Window.orderBy(F.col("GASTO_TOTAL").desc())
kpi_valor = kpi_valor.withColumn("RANK_VALOR", F.percent_rank().over(window_spec))

kpi_valor = kpi_valor.withColumn("SEGMENTO_VALOR",
    F.when(F.col("RANK_VALOR") <= 0.2, "A - ALTO VALOR")
    .when(F.col("RANK_VALOR") <= 0.5, "B - VALOR MEDIO")
    .otherwise("C - BAJO VALOR")
)

print(f" KPIs de valor creados: {kpi_valor.count():,} clientes")
print(" Columnas creadas:", kpi_valor.columns)
```

```python
# Visualizaciones de KPIs de valor económico

# Convertir a Pandas para visualización

kpi_valor_pd = kpi_valor.toPandas()

fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=('Segmentos', 'Gasto', 'Rentabilidad', 'Uso Límite', 'Productos', 'Top 10'),
    specs=[
        [{'type': 'domain'}, {'type': 'xy'}, {'type': 'xy'}],
        [{'type': 'xy'}, {'type': 'xy'}, {'type': 'xy'}]
    ]
)

# 1. Tarta
valor_counts = kpi_valor_pd['SEGMENTO_VALOR'].value_counts()
fig.add_trace(go.Pie(labels=valor_counts.index, values=valor_counts.values), row=1, col=1)

# 2. Barras
gasto_segmento = kpi_valor_pd.groupby('SEGMENTO_VALOR')['GASTO_TOTAL'].mean()
fig.add_trace(go.Bar(x=gasto_segmento.index, y=gasto_segmento.values), row=1, col=2)

# 3. Dispersión 
n_points = 500 if len(kpi_valor_pd) > 500 else len(kpi_valor_pd)
sample = kpi_valor_pd.sample(n=n_points)
fig.add_trace(go.Scatter(x=sample['TOTAL_INCOME'], y=sample['RENTABILIDAD_ESTIMADA'], mode='markers'), row=1, col=3)

# 4. Histograma
fig.add_trace(go.Histogram(x=kpi_valor_pd['PORCENTAJE_USO_LIMITE']), row=2, col=1)

# 5. Boxplot
fig.add_trace(go.Box(y=kpi_valor_pd['NUMBER_OF_PRODUCTS']), row=2, col=2)

# 6. Top 10
top_10 = kpi_valor_pd.nlargest(10, 'RENTABILIDAD_ESTIMADA')
fig.add_trace(go.Bar(x=top_10['CLIENT_ID'].astype(str), y=top_10['RENTABILIDAD_ESTIMADA']), row=2, col=3)

fig.update_layout(height=800, title_text="Valor Económico")
fig.show()
```

```python
#  3. KPIs de interacción y fidelidad

print("\n Calculando KPIs de interacción y fidelidad...")

# Interacción digital
kpi_digital = df_clients.select(
    "CLIENT_ID",
    "DIGITAL_CLIENT",
    "HOME_OWNER",
    "OWN_INSURANCE_CAR"
).withColumn(
    "CLIENTE_DIGITAL",
    when(col("DIGITAL_CLIENT") == 1, "SI").otherwise("NO")
)

# Comportamiento de fidelidad (retención)
# Calcular meses consecutivos con actividad
window_cliente = Window.partitionBy("CLIENT_ID").orderBy("YEAR", "MONTH")

# Primero agrupar por cliente y mes
df_mensual = df_behaviour.groupBy("CLIENT_ID", "YEAR", "MONTH", "YEAR_MONTH").agg(
    count("*").alias("TRANSACCIONES_MES")
).withColumn(
    "MES_ANTERIOR",
    lag("YEAR_MONTH", 1).over(window_cliente)
).withColumn(
    "ES_CONSECUTIVO",
    when(
        (col("YEAR") == lag("YEAR", 1).over(window_cliente)) & 
        (col("MONTH") - lag("MONTH", 1).over(window_cliente) == 1), 
        1
    ).otherwise(0)
)

# Calcular racha máxima de meses consecutivos
kpi_racha = df_mensual.groupBy("CLIENT_ID").agg(
    F.sum("ES_CONSECUTIVO").alias("MESES_CONSECUTIVOS"),
    count("*").alias("MESES_TOTALES")
).withColumn(
    "TASA_RETENCION",
    when(col("MESES_TOTALES") > 0,
         (col("MESES_CONSECUTIVOS") / col("MESES_TOTALES")) * 100)
    .otherwise(0)
)

# Unir todos los KPIs de fidelidad
kpi_fidelidad_final = kpi_digital.join(kpi_racha, on="CLIENT_ID", how="left")

# Segmentación por fidelidad
kpi_fidelidad_final = kpi_fidelidad_final.withColumn("SEGMENTO_FIDELIDAD",
    when(col("TASA_RETENCION") >= 80, "ALTA_FIDELIDAD")
    .when(col("TASA_RETENCION") >= 50, "MEDIA_FIDELIDAD")
    .otherwise("BAJA_FIDELIDAD")
).fillna({"TASA_RETENCION": 0, "SEGMENTO_FIDELIDAD": "SIN_HISTORIAL"})

print(f" KPIs de fidelidad calculados: {kpi_fidelidad_final.count():,} clientes")
```

```python
# Visualización de fidelidad

kpi_fidelidad_pd = kpi_fidelidad_final.toPandas()

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Clientes Digitales vs No Digitales',
                    'Distribución Tasa de Retención',
                    'Fidelidad por Segmento Digital'),
    specs=[[{'type': 'domain'}, {'type': 'histogram'}],
           [{'type': 'bar'}, {'type': 'scatter'}]]
)

# 1. Clientes digitales
digital_counts = kpi_fidelidad_pd['CLIENTE_DIGITAL'].value_counts()
fig.add_trace(
    go.Pie(labels=digital_counts.index, 
           values=digital_counts.values,
           hole=0.3,
           name='Clientes Digitales'),
    row=1, col=1
)

# 2. Histograma tasa de retención
fig.add_trace(
    go.Histogram(x=kpi_fidelidad_pd['TASA_RETENCION'].clip(0, 100),
                 nbinsx=20,
                 name='Tasa Retención',
                 marker_color='lightcoral'),
    row=1, col=2
)

# 3. Fidelidad por segmento digital
fidelidad_digital = kpi_fidelidad_pd.groupby(['CLIENTE_DIGITAL', 'SEGMENTO_FIDELIDAD']).size().unstack(fill_value=0)

# Asegurar que todas las categorías existan
categorias_fidelidad = ['ALTA_FIDELIDAD', 'MEDIA_FIDELIDAD', 'BAJA_FIDELIDAD', 'SIN_HISTORIAL']
for cat in categorias_fidelidad:
    if cat not in fidelidad_digital.columns:
        fidelidad_digital[cat] = 0

for segmento in fidelidad_digital.columns:
    fig.add_trace(
        go.Bar(x=fidelidad_digital.index, 
               y=fidelidad_digital[segmento], 
               name=segmento),
        row=2, col=1
    )

fig.show()

 
```

```python
# 4.  KPIs de riesgo

# KPI 4.1: Riesgo financiero
kpi_riesgo = df_clients.select(
    "CLIENT_ID",
    "REACTIVE_SCORING",
    "PROACTIVE_SCORING", 
    "BEHAVIORAL_SCORING",
    "NON_COMPLIANT_CONTRACT",
    "INSTALLMENT",
    "TOTAL_INCOME"
)

# Calcular score promedio y ratios de riesgo
kpi_riesgo = kpi_riesgo.withColumn(
    "SCORE_PROMEDIO",
    when(
        (col("REACTIVE_SCORING").isNotNull()) &
        (col("PROACTIVE_SCORING").isNotNull()) &
        (col("BEHAVIORAL_SCORING").isNotNull()),
        (col("REACTIVE_SCORING") + col("PROACTIVE_SCORING") + col("BEHAVIORAL_SCORING")) / 3
    ).otherwise(0)
).withColumn(
    "RATIO_DEUDA_INGRESO",
    when(col("TOTAL_INCOME") > 0, 
         (col("INSTALLMENT") / col("TOTAL_INCOME")) * 100)
    .otherwise(100)
).fillna({"SCORE_PROMEDIO": 0, "RATIO_DEUDA_INGRESO": 100})

# KPI 4.2: Riesgo de abandono (churn) - unir con actividad y fidelidad
kpi_churn = kpi_actividad.select(
    "CLIENT_ID",
    "DIAS_DESDE_ULTIMA_ACTIVIDAD",
    "SEGMENTO_ACTIVIDAD"
).join(
    kpi_fidelidad_final.select("CLIENT_ID", "TASA_RETENCION"),
    on="CLIENT_ID",
    how="inner"
)

# Calcular probabilidad de churn (simplificado)
kpi_churn = kpi_churn.withColumn("PROBABILIDAD_CHURN",
    when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") > 180, 0.9)
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") > 90, 0.7)
    .when(col("DIAS_DESDE_ULTIMA_ACTIVIDAD") > 30, 0.4)
    .when(col("TASA_RETENCION") < 30, 0.6)
    .otherwise(0.2)
)

# Unir todos los KPIs de riesgo
kpi_riesgo_final = kpi_riesgo.join(kpi_churn, on="CLIENT_ID", how="inner")

# Segmentación por riesgo
kpi_riesgo_final = kpi_riesgo_final.withColumn("NIVEL_RIESGO",
    when(col("PROBABILIDAD_CHURN") >= 0.7, "ALTO_RIESGO")
    .when(col("PROBABILIDAD_CHURN") >= 0.4, "MEDIO_RIESGO")
    .when(col("NON_COMPLIANT_CONTRACT") == 1, "RIESGO_CONTRATO")
    .when(col("RATIO_DEUDA_INGRESO") > 40, "RIESGO_FINANCIERO")
    .otherwise("BAJO_RIESGO")
)

print(f" KPIs de riesgo calculados: {kpi_riesgo_final.count():,} clientes")
```

```python
# Visualización de riesgo

kpi_riesgo_pd = kpi_riesgo_final.toPandas()

fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=('Distribución Niveles de Riesgo',
                    'Probabilidad de Churn',
                    'Score Promedio vs Deuda/Ingreso',
                    'Riesgo por Segmento de Actividad',
                    'Clientes con Contrato No Cumplido',
                    'Top 10 Clientes de Alto Riesgo'),
    specs=[[{'type': 'domain'}, {'type': 'histogram'}, {'type': 'scatter'}],
           [{'type': 'bar'}, {'type': 'bar'}, {'type': 'bar'}]]
)

# 1. Distribución niveles de riesgo
riesgo_counts = kpi_riesgo_pd['NIVEL_RIESGO'].value_counts()
fig.add_trace(
    go.Pie(labels=riesgo_counts.index, 
           values=riesgo_counts.values,
           hole=0.3,
           name='Niveles Riesgo'),
    row=1, col=1
)

# 2. Histograma probabilidad de churn
fig.add_trace(
    go.Histogram(x=kpi_riesgo_pd['PROBABILIDAD_CHURN'],
                 nbinsx=20,
                 name='Probabilidad Churn',
                 marker_color='salmon'),
    row=1, col=2
)

# 3. Dispersión score vs ratio deuda/ingreso
muestra_riesgo = kpi_riesgo_pd.sample(min(1000, len(kpi_riesgo_pd)), random_state=42)
fig.add_trace(
    go.Scatter(x=muestra_riesgo['SCORE_PROMEDIO'],
               y=muestra_riesgo['RATIO_DEUDA_INGRESO'],
               mode='markers',
               marker=dict(
                   size=10,
                   color=muestra_riesgo['PROBABILIDAD_CHURN'],
                   colorscale='RdYlGn_r',
                   showscale=True,
                   colorbar=dict(title="Prob. Churn", x=1.0)
               ),
               text=muestra_riesgo['NIVEL_RIESGO'],
               name='Clientes'),
    row=1, col=3
)

# 4. Riesgo por segmento de actividad
riesgo_actividad = kpi_riesgo_pd.groupby('SEGMENTO_ACTIVIDAD')['PROBABILIDAD_CHURN'].mean()
fig.add_trace(
    go.Bar(x=riesgo_actividad.index, 
           y=riesgo_actividad.values,
           name='Prob. Churn Promedio',
           marker_color='coral'),
    row=2, col=1
)

# 5. Clientes con contrato no cumplido
contrato_no_cumplido = kpi_riesgo_pd[kpi_riesgo_pd['NON_COMPLIANT_CONTRACT'] == 1]
if len(contrato_no_cumplido) > 0:
    fig.add_trace(
        go.Bar(x=['Contrato No Cumplido'], 
               y=[len(contrato_no_cumplido)],
               name='Contrato No Cumplido',
               marker_color='darkred'),
        row=2, col=2
    )
else:
    fig.add_trace(
        go.Bar(x=['Sin Datos'], y=[0]),
        row=2, col=2
    )

# 6. Top 10 clientes de alto riesgo
alto_riesgo = kpi_riesgo_pd[kpi_riesgo_pd['NIVEL_RIESGO'] == 'ALTO_RIESGO']
if len(alto_riesgo) > 0:
    top_10_riesgo = alto_riesgo.nlargest(10, 'PROBABILIDAD_CHURN')
    fig.add_trace(
        go.Bar(x=top_10_riesgo['CLIENT_ID'].astype(str),
               y=top_10_riesgo['PROBABILIDAD_CHURN'],
               name='Alto Riesgo',
               marker_color='red'),
        row=2, col=3
    )
else:
    fig.add_trace(
        go.Bar(x=['Sin Datos'], y=[0]),
        row=2, col=3
    )

fig.update_layout(
    height=800, 
    title_text=" KPIs de Riesgo Potencial - FinPlus", 
    showlegend=False,
    template='plotly_white'
)

# Mejorar ejes
fig.update_xaxes(title_text="Segmento Actividad", row=2, col=1)
fig.update_xaxes(title_text="Probabilidad Churn", row=1, col=2)
fig.update_xaxes(title_text="Score Promedio", row=1, col=3)

fig.update_yaxes(title_text="Frecuencia", row=1, col=2)
fig.update_yaxes(title_text="Ratio Deuda/Ingreso (%)", row=1, col=3)
fig.update_yaxes(title_text="Prob. Churn Promedio", row=2, col=1)

fig.show()
```

```python
# 5. KPI de oportunidades comerciales

# Unir todos los KPIs para análisis cruzado
kpi_completo = kpi_actividad \
    .join(kpi_valor, on="CLIENT_ID", how="inner") \
    .join(kpi_fidelidad_final.select("CLIENT_ID", "CLIENTE_DIGITAL", "TASA_RETENCION", "OWN_INSURANCE_CAR"), 
          on="CLIENT_ID", how="inner") \
    .join(kpi_riesgo_final.select("CLIENT_ID", "SCORE_PROMEDIO", "RATIO_DEUDA_INGRESO", "NON_COMPLIANT_CONTRACT"), 
          on="CLIENT_ID", how="inner")

# Añadir CAR_AGE desde df_clients si existe
if "CAR_AGE" in df_clients.columns:
    car_age_df = df_clients.select("CLIENT_ID", "CAR_AGE")
    kpi_completo = kpi_completo.join(car_age_df, on="CLIENT_ID", how="left")
else:
    kpi_completo = kpi_completo.withColumn("CAR_AGE", lit(None))

# KPI 5.1: Oportunidades de cross-selling
kpi_oportunidades = kpi_completo.withColumn("OPORTUNIDAD_CROSS_SELLING",
    when((col("NUMBER_OF_PRODUCTS") == 1) & 
         (col("SEGMENTO_VALOR").isin(["A - ALTO VALOR", "B - VALOR MEDIO"])), 
         "ALTA_OPORTUNIDAD")
    .when((col("NUMBER_OF_PRODUCTS") <= 2) & 
          (col("TOTAL_INCOME") > kpi_completo.agg(mean("TOTAL_INCOME")).collect()[0][0]), 
          "MEDIA_OPORTUNIDAD")
    .otherwise("BAJA_OPORTUNIDAD")
)

# KPI 5.2: Oportunidades de engagement (re-engagement)
kpi_oportunidades = kpi_oportunidades.withColumn("OPORTUNIDAD_ENGAGEMENT",
    when((col("SEGMENTO_ACTIVIDAD") == "INACTIVO") & 
         (col("SEGMENTO_VALOR") == "A - ALTO VALOR"), 
         "RECUPERACION_ALTO_VALOR")
    .when((col("SEGMENTO_ACTIVIDAD").isin(["POCO_ACTIVO", "INACTIVO"])) & 
          (col("CLIENTE_DIGITAL") == "NO"), 
          "MIGRACION_DIGITAL")
    .when((col("OWN_INSURANCE_CAR") == "N") & 
          (col("CAR_AGE").isNotNull()) & (col("CAR_AGE") > 0), 
          "SEGURO_AUTOMOVIL")
    .otherwise("MANTENIMIENTO")
)

# KPI 5.3: Clientes ideales para up-selling
kpi_oportunidades = kpi_oportunidades.withColumn("OPORTUNIDAD_UP_SELLING",
    when((col("PORCENTAJE_USO_LIMITE") > 70) & 
         (col("SCORE_PROMEDIO") > 70), 
         "AUMENTO_LIMITE_CREDITO")
    .when((col("RENTABILIDAD_ESTIMADA") > 0) & 
          (col("NUMBER_OF_PRODUCTS") < 3), 
          "PRODUCTOS_PREMIUM")
    .otherwise("NO_APLICA")
)

print(f"KPIs de oportunidades calculados: {kpi_oportunidades.count():,} clientes")
```

```python
# RESUMEN

# Calcular métricas resumen
resumen_kpis = {}

# Métricas básicas
resumen_kpis["TOTAL_CLIENTES"] = kpi_oportunidades.count()

# Actividad
resumen_kpis["CLIENTES_ACTIVOS"] = kpi_oportunidades.filter(col("SEGMENTO_ACTIVIDAD") == "ACTIVO").count()
resumen_kpis["CLIENTES_INACTIVOS"] = kpi_oportunidades.filter(col("SEGMENTO_ACTIVIDAD") == "INACTIVO").count()
resumen_kpis["TASA_ACTIVIDAD_PROMEDIO"] = kpi_oportunidades.agg(mean("TOTAL_TRANSACCIONES")).collect()[0][0]

# Valor
resumen_kpis["CLIENTES_ALTO_VALOR"] = kpi_oportunidades.filter(col("SEGMENTO_VALOR") == "A - ALTO VALOR").count()
resumen_kpis["GASTO_TOTAL_PROMEDIO"] = kpi_oportunidades.agg(mean("GASTO_TOTAL")).collect()[0][0]
resumen_kpis["RENTABILIDAD_PROMEDIO"] = kpi_oportunidades.agg(mean("RENTABILIDAD_ESTIMADA")).collect()[0][0]

# Fidelidad
resumen_kpis["CLIENTES_DIGITALES"] = kpi_oportunidades.filter(col("CLIENTE_DIGITAL") == "SI").count()
resumen_kpis["TASA_RETENCION_PROMEDIO"] = kpi_oportunidades.agg(mean("TASA_RETENCION")).collect()[0][0]

# Riesgo
resumen_kpis["CLIENTES_ALTO_RIESGO"] = kpi_oportunidades.filter(col("NIVEL_RIESGO") == "ALTO_RIESGO").count()
resumen_kpis["CONTRATOS_NO_CUMPLIDOS"] = kpi_oportunidades.filter(col("NON_COMPLIANT_CONTRACT") == 1).count()

# Oportunidades
resumen_kpis["OPORTUNIDADES_CROSS_SELLING"] = kpi_oportunidades.filter(
    col("OPORTUNIDAD_CROSS_SELLING") == "ALTA_OPORTUNIDAD"
).count()

resumen_kpis["OPORTUNIDADES_ENGAGEMENT"] = kpi_oportunidades.filter(
    col("OPORTUNIDAD_ENGAGEMENT").isin(["RECUPERACION_ALTO_VALOR", "MIGRACION_DIGITAL", "SEGURO_AUTOMOVIL"])
).count()

# Mostrar resumen
print("\n" + "="*70)
print(" RESUMEN EJECUTIVO - FINPLUS ANALYTICS CHALLENGE")
print("="*70)

print("\n POBLACIÓN DE CLIENTES:")
print(f"   • Total Clientes: {resumen_kpis['TOTAL_CLIENTES']:,}")
print(f"   • Clientes Activos: {resumen_kpis['CLIENTES_ACTIVOS']:,} ({resumen_kpis['CLIENTES_ACTIVOS']/resumen_kpis['TOTAL_CLIENTES']*100:.1f}%)")
print(f"   • Clientes Digitales: {resumen_kpis['CLIENTES_DIGITALES']:,} ({resumen_kpis['CLIENTES_DIGITALES']/resumen_kpis['TOTAL_CLIENTES']*100:.1f}%)")

print("\n VALOR ECONÓMICO:")
print(f"   • Clientes Alto Valor (Segmento A): {resumen_kpis['CLIENTES_ALTO_VALOR']:,}")
print(f"   • Gasto Promedio por Cliente: €{resumen_kpis['GASTO_TOTAL_PROMEDIO']:,.2f}")
print(f"   • Rentabilidad Promedio: €{resumen_kpis['RENTABILIDAD_PROMEDIO']:,.2f}")

print("\n RIESGO POTENCIAL:")
print(f"   • Clientes Alto Riesgo: {resumen_kpis['CLIENTES_ALTO_RIESGO']:,}")
print(f"   • Contratos No Cumplidos: {resumen_kpis['CONTRATOS_NO_CUMPLIDOS']:,}")

print("\n OPORTUNIDADES COMERCIALES:")
print(f"   • Cross-Selling (Alta Oportunidad): {resumen_kpis['OPORTUNIDADES_CROSS_SELLING']:,}")
print(f"   • Re-engagement: {resumen_kpis['OPORTUNIDADES_ENGAGEMENT']:,}")

print("\n KPIs CLAVE:")
print(f"   • Tasa Actividad Promedio: {resumen_kpis['TASA_ACTIVIDAD_PROMEDIO']:.1f} transacciones/cliente")
print(f"   • Tasa Retención Promedio: {resumen_kpis['TASA_RETENCION_PROMEDIO']:.1f}%")
print(f"   • Inactividad: {resumen_kpis['CLIENTES_INACTIVOS']:,} clientes")

print("\n INSIGHTS PRINCIPALES:")
print("   1. Segmentar campañas por valor (A, B, C)")
print("   2. Priorizar recuperación de clientes inactivos de alto valor")
print("   3. Incentivar migración digital entre clientes no digitales")
print("   4. Monitorear clientes de alto riesgo con scoring < 50")

print("="*70)
```
